## Segunda parte

En esta segunda parte vamos a tratar con conjuntos de datos cuyos atributos no son numéricos. Puesto que el perceptrón necesita que los valores de los atributos sean numéricos, en esos casos es habitual aplicar una transformación conocida como _codificación "one hot"_.

En la codificación "one hot" cada atributo se sustituye por $n$ atributos numéricos, donde $n$ es el número de valores distintos que puede tomar el atributo. En general, si $A$ es un atributo que puede tomar valores $v_1,\ldots,v_n$, entonces el valor $v_i$ se sustituye por los valores $(0,0,\ldots,0,1,0,\ldots,0)$  donde hay un 1 en la posición $i$-ésima y 0 en el resto.

Esta codificación se aplica con todos los atributos, concatenando las tuplas obtenidas. Por ejemplo, si los valores de cada columna se consideran en el siguiente orden:

* Primer atributo: _Lluvia_, _Nublado_, _Soleado_

La siguiente función `lista_valores(datos)` obtiene una lista cuyos elementos son las listas de valores (ordenados alfabéticamente) de cada columna del conjunto de datos `datos`.

```python
def lista_valores(entr):
    return [sorted(list(set((ej[i] for ej in entr)))) for i in range(len(entr[0]))]
```
